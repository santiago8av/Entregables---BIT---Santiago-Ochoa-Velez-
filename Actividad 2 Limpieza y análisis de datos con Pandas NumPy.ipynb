### 3. NumPy
Utiliza NumPy para realizar transformaciones sobre el DataFrame.
* Crea una variable utilizando:
```python
np.where()
```
La variable debe representar una condición relevante para el contexto de los datos.
* Utiliza al menos una función estadística de NumPy para calcular una medida sobre una variable numérica.

Puedes utilizar, por ejemplo:

```python
np.mean()
np.median()
np.nanmean()
np.nanmedian()
np.std()
```
Explica brevemente qué representa el resultado obtenido.

````python
mediapeso = df.groupby("Animal type")["Weight kg"].transform(lambda x: np.nanmean(x))
stdpeso   = df.groupby("Animal type")["Weight kg"].transform(lambda x: np.nanstd(x))

mediatam  = df.groupby("Animal type")["Body Length cm"].transform(lambda x: np.nanmean(x))
stdtam    = df.groupby("Animal type")["Body Length cm"].transform(lambda x: np.nanstd(x))
LStam = mediatam + 2 * stdtam
LSpes = mediapeso + 2 * stdpeso
LItam = mediatam - 2 * stdtam
LIpes = mediapeso - 2 * stdpeso
df["Weight_Category"] = np.where(
    (df["Weight kg"] > LSpes) | (df["Weight kg"] < LIpes), 
    "Atípico", 
    "Normal"
)
df["Length_Category"] = np.where(
    (df["Body Length cm"] > LStam) | (df["Body Length cm"] < LItam), 
    "Atípico", 
    "Normal"
)
print(df["Weight_Category"].value_counts())
print(df["Length_Category"].value_counts())
````

La idea de estas categorias es identicar animales que estan fuera del promedio de tamaño o peso para el animal segun el estudio
